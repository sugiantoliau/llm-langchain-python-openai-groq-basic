### Initialization AI  using OpenAI model GPT-4o mini

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

llm = ChatOpenAI(
    model="gpt-4o-mini", 
    temperature=0.7,
)
```

### Create the Algoritm

```python
# Create a Prompt Template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a senior AI assistant in an automotive company."),
    ("user", "Briefly explain the function of the {component} component in a car.")
])

# Connect the Template and LLM into a single pipeline (Chain)
chain = prompt | llm

```

```python
# Run the AI by passing the desired variables
# Start play here by changing "Carburetor"  to other words or try: "Turbocharger" or try: "ECU (Engine Control Unit)"

print("Asking the AI...")
respons = chain.invoke({"component": "Carburetor"})
```

```python
# Display the AI's response
print("\nAI Response:")
print(respons.content)
```
