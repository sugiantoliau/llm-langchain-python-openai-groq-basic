### Initialization AI  using Grog (Meta Llama 3.1 8B model)

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

llm = ChatGroq(
    model="llama-3.1-8b-instant",  # Updated to the currently active model
    temperature=0.7,
)

```

### Create the Algoritm

```python
# Create a Prompt Template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a senior AI assistant in an automotive company. Respond in English."),
    ("user", "Briefly explain the function of the {component} component in a car.")
])

# Connect the components into a single pipeline (Chain)
chain = prompt | llm
```

```python
# Run the AI by passing the desired variables
# Start play here by changing "Carburetor"  to other words or try: "Turbocharger" or try: "ECU (Engine Control Unit)"

print("Asking Llama 3 via Groq (Free)...")
respons = chain.invoke({"component": "Turbocharger"})
```

```python
# 5. Display the AI's response
print("\nAI Response:")
print(respons.content)
```
